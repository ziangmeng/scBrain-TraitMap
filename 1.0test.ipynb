```python
from pathlib import Path
import anndata as ad
import numpy as np

# 原始文件路径
rna_path  = Path("/io/mza/glue/34gw/data/rna_stratified_100_per_sid_sc.h5ad")
atac_path = Path("/io/mza/glue/34gw/data/adata_brain_filtered.noEC.pp.h5ad")

# 输出文件夹（当前目录 /data）
out_dir = Path("./data")
out_dir.mkdir(exist_ok=True)

def subsample_and_save(path, out_name, frac=0.01, seed=42):
    print(f">>> Loading {path}")
    adata = ad.read_h5ad(path)

    n = adata.n_obs
    k = max(1, int(n * frac))

    print(f"Total cells = {n}, sampling = {k} cells (fraction={frac})")
    idx = np.random.RandomState(seed).choice(n, size=k, replace=False)

    adata_sub = adata[idx].copy()

    out_path = out_dir / out_name
    adata_sub.write_h5ad(out_path, compression="gzip")

    print(f"Saved subsampled data → {out_path}")
    return out_path

rna_out  = subsample_and_save(rna_path,  "rna_subsampled.h5ad",  frac=0.01)
atac_out = subsample_and_save(atac_path, "atac_subsampled.h5ad", frac=0.01)

print("\nAll done ✓")
print("RNA saved at :", rna_out)
print("ATAC saved at:", atac_out)
```
